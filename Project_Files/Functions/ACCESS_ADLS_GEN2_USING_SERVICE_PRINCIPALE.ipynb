```python
application_id = dbutils.secrets.get(scope="access-key-from-kv", key="application-id")
directory_id = dbutils.secrets.get(scope="access-key-from-kv", key="directory-id")
secret = dbutils.secrets.get(scope="access-key-from-kv", key="secret")

spark.conf.set("fs.azure.account.auth.type.retailstorage1881.dfs.core.windows.net", "OAuth")
spark.conf.set("fs.azure.account.oauth.provider.type.retailstorage1881.dfs.core.windows.net", "org.apache.hadoop.fs.azurebfs.oauth2.ClientCredsTokenProvider")
spark.conf.set("fs.azure.account.oauth2.client.id.retailstorage1881.dfs.core.windows.net", application_id)
spark.conf.set("fs.azure.account.oauth2.client.secret.retailstorage1881.dfs.core.windows.net", secret)
spark.conf.set("fs.azure.account.oauth2.client.endpoint.retailstorage1881.dfs.core.windows.net", f"https://login.microsoftonline.com/{directory_id}/oauth2/token")
```
