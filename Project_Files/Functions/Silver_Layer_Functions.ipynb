```python
from pyspark.sql.functions import *
from pyspark.sql.types import StringType
```

```python
# Rename columns (snake_case)
def snake_case(df):
    return df.toDF(
        *[c.strip().lower().replace(' ', '_') for c in df.columns]
    )
```

```python
# Remove spaces from column values
def trim_col_values(df):
    return df.select([trim(col(c)).alias(c)
        if dict(df.dtypes)[c] == 'string'
        else col(c)
        for c in df.columns
    ])
```

```python
# Cast data types
def cast_dtype(df, column_dtype_map):
    return df.select(
        *[
            expr(
                f"try_cast({column} as {datatype})"
            ).alias(column)

            for column, datatype in column_dtype_map.items()
        ]
    )
```

```python
# Remove spaces from a column in a PySpark DataFrame

def remove_spaces(df, column_name):
    return df.withColumn(
        column_name,
        regexp_replace(col(column_name), " ", "")
    )
```

```python
# lower case column
def lower_case(df, column_name):
    return df.withColumn(
        column_name,
        lower(col(column_name))
    )
```

```python
# initcap column
def initcap_column(df, column_name):
    return df.withColumn(
        column_name,
        initcap(col(column_name))
    )
```

```python
# Remove $ symbol
def remove_dollar_symbol(df, column_name):
    return df.withColumn(
    column_name,
    regexp_replace(col(column_name), "\\$", "").cast("double")
)
```

```python
# Remove special characters
def remove_null_and_special_chars(df, column_name):
    return df.withColumn(
        column_name,
        expr(
            f"try_cast(regexp_replace({column_name}, '[^0-9.]', '') as double)"
        )
    )
```

```python
# Remove null and special characters
def remove_special_chars(df, column_name):
    return  df.withColumn(
    column_name,
    regexp_replace(col(column_name), "[@#]", "_")
)
```
